**Question 1: RDD – Filter and Transform Numbers**

```python
rdd = sc.parallelize([5, 12, 7, 20, 3, 16, 9, 24, 10, 1])
result = (
    rdd
    .filter(lambda x: x > 10)
    .map(lambda x: x * 2)
)
print(result.collect())
```

**Question 2: RDD – Student Marks**

```python
rdd = sc.parallelize([
    ("Arun", 78),
    ("Divya", 92),
    ("Kavin", 45),
    ("Meena", 66),
    ("Ravi", 35),
    ("Priya", 88)
])
result = (
    rdd
    .filter(lambda x: x[1] >= 50)
    .map(lambda x: f"{x[0]} scored {x[1]}")
)

print(result.collect())
```

Question 3: DataFrame – Select and Filter Employees

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

sparkSession = SparkSession.builder.appName("DataFrameAssig").master("local[*]").getOrCreate()
print(sparkSession)
sc = sparkSession.sparkContext
```

```python
employee_data = [
    (1, "Arun", "Male", 24, "Chennai", 28000),
    (2, "Divya", "Female", 27, "Bengaluru", 42000),
    (3, "Kavin", "Male", 22, "Chennai", 25000),
    (4, "Meena", "Female", 30, "Hyderabad", 52000),
    (5, "Ravi", "Male", 28, "Bengaluru", 38000),
    (6, "Priya", "Female", 25, "Chennai", 46000)
]

columns = ["id", "name", "gender", "age", "city", "salary"]
emp_df = sparkSession.createDataFrame(employee_data, columns)
emp_df.show()
```

```python
emp_df.select("name", "city", "salary").show()
```

```python
emp_df.filter(emp_df.salary > 40000).show()
```

**Question 4: DataFrame – Add a Column Using "withColumn()"**



```python
product_data = [
    (101, "Notebook", 40, 5),
    (102, "Pen", 10, 20),
    (103, "Pencil", 5, 30),
    (104, "School Bag", 650, 3),
    (105, "Eraser", 8, 15)
]

columns = ["product_id", "product_name", "price", "quantity"]
prd_df = sparkSession.createDataFrame(product_data, columns)
prd_df.show()
```

```python
from pyspark.sql.functions import col
prd_df = prd_df.withColumn("total_value",col("price") * col("quantity"))
prd_df.show()
```

Question 5: DataFrame – Filter, Select, and Add a Column



```python
movie_data = [
    ("Leo", "Action", 250, 7.8),
    ("Hi Nanna", "Drama", 75, 8.2),
    ("Jailer", "Action", 200, 7.1),
    ("Good Night", "Comedy", 30, 8.0),
    ("Dragon", "Comedy", 100, 7.5),
    ("Amaran", "Drama", 150, 8.3)
]

columns = ["movie_name", "category", "collection_crore", "rating"]
movie_df = sparkSession.createDataFrame(movie_data, columns)
movie_df.select("movie_name", "category", "rating").show()
```

```python
high_rated_movies = movie_df.filter(col("rating") >= 8.0)

df2 = high_rated_movies.withColumn(
    "collection_with_10_percent_bonus",
    col("collection_crore") * 1.10
)
df2.show()
```
