````python
data = [
    (101, "Arun", "IT", 45000, 2, "Chennai"),
    (102, "Priya", "HR", 38000, 3, "Bengaluru"),
    (103, "Karthik", "IT", 52000, 5, "Chennai"),
    (104, "Divya", "Finance", 41000, 1, "Hyderabad"),
    (105, "Rahul", "Sales", 35000, 2, "Pune"),
    (106, "Meena", "IT", 60000, 6, "Bengaluru"),
    (107, "Vijay", "Sales", 47000, 4, "Chennai"),
    (108, "Sneha", "HR", 43000, 2, "Pune"),
    (109, "Ajay", "Finance", 56000, 7, "Hyderabad"),
    (110, "Nila", "IT", 39000, 1, "Chennai"),
    (111, "Manoj", "Sales", 50000, 5, "Bengaluru"),
    (112, "Keerthi", "HR", 46000, 4, "Hyderabad")
]

columns = [
    "employee_id",
    "employee_name",
    "department",
    "salary",
    "experience_years",
    "city"
]

df = spark.createDataFrame(data, columns)
df.show()
df.printSchema()
````

### Question 2 --- Create an Annual Salary Column

Using `withColumn()`:

1.  Create a column named `annual_salary`.
2.  Calculate annual salary using the monthly `salary` multiplied by
    `12`.
3.  Display only `employee_name`, `salary`, and `annual_salary`.

**Expected concepts:** `withColumn()` and `select()`.


### Question 3 --- Add a Fixed Company Column

Using `withColumn()` and `lit()`:

1.  Add a column named `company_name`.
2.  Set its value to `"Vertex Automation Labs"` for every employee.
3.  Display the employee name, department, and company name.

### Question 4 --- Create an Experience Category

Using `withColumn()` and `when()` / `otherwise()`:

Create a column named `experience_category` using these rules:

-   Less than 2 years: `"Fresher"`
-   From 2 through 4 years: `"Mid-Level"`
-   More than 4 years: `"Experienced"`

Display `employee_name`, `experience_years`, and `experience_category`.

**Expected concepts:** `withColumn()`, `when()`, and `otherwise()`.

````python
df.withColumn(
    "experience_category",
    when(
        col("experience_years") < 2,"Fresher")
    .when(
        col("experience_years")  <= 4,"Mid-Level")
    .otherwise("Experienced")
).select(
    "employee_name",
    "experience_years",
    "experience_category"
).show()
````

### Question 5 --- Filter and Transform Employee Data

Perform the following transformations:

1.  Create a column named `salary_with_bonus` by adding ₹5,000 to each
    employee's salary.
2.  Filter the DataFrame to include only employees whose
    `salary_with_bonus` is greater than ₹50,000.
3.  Select `employee_name`, `department`, `salary`, and
    `salary_with_bonus`.
4.  Sort the result by `salary_with_bonus` in descending order.

**Expected concepts:** `withColumn()`, arithmetic expressions,
`filter()` / `where()`, `select()`, and `orderBy()`.

````python
df.withColumn(
    "salary_with_bonus",
    col("salary") + 5000
).filter(
    col("salary_with_bonus") > 50000
).select(
    "employee_name",
    "department",
    "salary",
    "salary_with_bonus"
).orderBy(
    col("salary_with_bonus").desc()
).show()
````
